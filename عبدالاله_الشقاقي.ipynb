```python
import platform
import os
import psutil
from datetime import datetime

# ==========================================
# Cloud Computing Project - Student Details
# ==========================================
STUDENT_NAME = "عبدالاله الشقاقي"
COURSE_NAME = "Cloud Computing - الحوسبة السحابية"
PROJECT_TITLE = "Cloud Instance Environment Diagnostic & Dashboard"

def print_header():
    print("=" * 65)
    print(f" ☁️  {PROJECT_TITLE}  ☁️")
    print("=" * 65)
    print(f" 👤 الطالب: {STUDENT_NAME}")
    print(f" 📚 المادة : {COURSE_NAME}")
    print(f" ⏰ التاريخ : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("=" * 65)
    print()

def get_cloud_specs():
    print(" 🚀 [معلومات الخادم السحابي - Google Colab VM Specs]")
    print("-" * 65)

    # OS Info
    print(f" • نظام التشغيل السحابي : {platform.system()} {platform.release()} ({platform.machine()})")
    print(f" • اسم المضيف (Hostname)  : {platform.node()}")

    # CPU Info
    print(f" • عدد أنوية المعالج (CPU) : {os.cpu_count()} Cores")

    # Memory Info
    ram_gb = round(psutil.virtual_memory().total / (1024**3), 2)
    ram_usage = psutil.virtual_memory().percent
    print(f" • الذاكرة العشوائية (RAM) : {ram_gb} GB (المستخدم: {ram_usage}%)")

    # Disk Info
    disk_gb = round(psutil.disk_usage('/').total / (1024**3), 2)
    disk_free = round(psutil.disk_usage('/').free / (1024**3), 2)
    print(f" • المساحة التخزينية المتاحة : {disk_free} GB / {disk_gb} GB")
    print("-" * 65)

def run_cloud_simulation():
    print("\n ⚡ [حالة الخدمة السحابية - Service Status]")
    print(" [" + "█" * 30 + "] 100% Completed")
    print(f" ✅ تم تشغيل البيئة السحابية بنجاح باسم المهندس/ {STUDENT_NAME}")
    print("=" * 65)

# تنفيذ البرنامج
if __name__ == "__main__":
    print_header()
    get_cloud_specs()
    run_cloud_simulation()
```
